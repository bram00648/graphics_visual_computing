```python
import numpy as np
import cv2

image = cv2.imread("landscape_downscaled.png")

cv2.imshow("Original Image", image)
cv2.waitKey(0)

gaussianBlur = cv2.GaussianBlur(image, (31, 31), 0)

# horizontal gausian blur
kernel = cv2.getGaussianKernel(31, 0).T
gaussianBlurHorizontal = cv2.filter2D(image, -1, kernel)

# Box filter
kernel = np.ones ((31, 31), np.float32)
kernel /= kernel.sum()
boxFilter = cv2.filter2D(image, -1, kernel)

# deduplication filter
kernel = np.zeros((31, 31), np.float32)
kernel[0,0] = 0.5
kernel[30,30] = 0.5
deduplicationFilter = cv2.filter2D(image, -1, kernel)


cv2.imshow("Gaussian Blur", gaussianBlur)
cv2.waitKey(0)

cv2.imshow("Gaussian Blur Horizontal", gaussianBlurHorizontal)
cv2.waitKey(0)

cv2.imshow("Box Filter", boxFilter)
cv2.waitKey(0)

cv2.imshow("Deduplication Filter", deduplicationFilter)
cv2.waitKey(0)
```
